```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import LabelEncoder, StandardScaler
print('Libraries imported.')
```

```python
url = '/content/bmw.csv'
desired_columns = ['model','year','price','transmission','fuelType','tax','mpg','engineSize']
# Load the data, letting pandas infer the header. This corrects the initial data loading issue.
df = pd.read_csv(url)
# Filter the DataFrame to keep only the desired columns.
# This handles cases where the CSV might contain additional columns (like 'mileage') not in our desired list.
df = df[desired_columns]
df.head()
```

```python
print('Shape:', df.shape)
print('\nColumn Names:', df.columns.tolist())
df.head()
```

```python
for col in df.select_dtypes(include=['float64','int64']).columns:
    df[col] = df[col].fillna(df[col].mean())


for col in df.select_dtypes(include=['object']).columns:
    df[col] = df[col].fillna(df[col].mode()[0])

df.isnull().sum()
```

```python
before = df.shape[0]
df.drop_duplicates(inplace=True)
after = df.shape[0]
print('Duplicates removed:', before - after)
```

```python
encoder = LabelEncoder()
df['model'] = encoder.fit_transform(df['model'])
df.head()
```
