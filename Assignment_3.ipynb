```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

print("Libraries imported.")
```

```python
url = "https://raw.githubusercontent.com/jbrownlee/Datasets/master/pima-indians-diabetes.data.csv"

columns = [
    'pregnancies', 'glucose', 'blood_pressure', 'skin_thickness',
    'insulin', 'bmi', 'diabetes_pedigree', 'age', 'class'
]

df = pd.read_csv(url, names=columns)
df.head()
```

```python
X = df.iloc[:, 0:8]   # Feature columns
y = df['class']      # Target column

print(X.shape, y.shape)
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42
)

print("Train size:", X_train.shape)
print("Test size:", X_test.shape)
```

```python
knn = KNeighborsClassifier(n_neighbors=5)
knn.fit(X_train, y_train)

print("Model trained.")
```

```python
y_pred = knn.predict(X_test)
list(y_pred)[:10]
```

```python
results = pd.DataFrame({
    'Actual': y_test.values,
    'Predicted': y_pred
})

results['Status'] = results.apply(
    lambda r: 'Correct' if r.Actual == r.Predicted else 'Wrong',
    axis=1
)

results.head(20)
```

```python
print("Accuracy:", accuracy_score(y_test, y_pred))

print("\nClassification Report:")
print(classification_report(y_test, y_pred))

print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_pred))
```

 Conclusion

The k-Nearest Neighbour algorithm was successfully applied to the Diabetes dataset. The model predicted whether a patient is diabetic based on medical diagnostic features such as glucose level, BMI, and age. The accuracy score and classification report indicate the performance of the classifier. Correct and incorrect predictions were analyzed using a results table and confusion matrix. This experiment demonstrates the effectiveness of kNN in healthcare-related binary classification problems.
